```python
import pandas as pd

# Load original dataset
df = pd.read_csv("sales_data.csv", encoding="latin1")

# Keep only relevant columns (Order ID added)
df_dashboard = df[
    [
        "Order ID",
        "Order Date",
        "Region",
        "Category",
        "Sub-Category",
        "Sales",
        "Quantity",
        "Profit"
    ]
]

# Convert 'Order Date' to datetime
df_dashboard["Order Date"] = pd.to_datetime(df_dashboard["Order Date"])

# Save dashboard-ready CSV
df_dashboard.to_csv("dashboard_sales_data.csv", index=False)

print("Dashboard-ready CSV saved!")
```
